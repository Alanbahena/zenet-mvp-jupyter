```python
import gradio as gr
from gradio_app.components import render_chat_panel

def fake_chat(message: str, history: list, sid: str):
    history = list(history)
    history.append({"role": "user", "content": message})
    history.append({"role": "assistant", "content": "Hola, como estas?"})
    return history, ""

```

```python
greeting = [{"role": "assistant", "content": "Hola, soy el asistente de prueba. ¿En qué puedo ayudarte?"}]
with gr.Blocks(title="Test render_chat_panel") as demo:
    session_id = gr.State('notebook-session')
    with gr.Column():
        render_chat_panel(
            chat_fn=fake_chat,
            session_id=session_id,
            data_lake_ref=None,
            initial_messages=greeting
        )
```

```python
demo.launch(inline=True)
```
